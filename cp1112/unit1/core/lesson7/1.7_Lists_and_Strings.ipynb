# 1.7 — Lists & Strings as Sequences

Every variable so far has held exactly one value. A **list** holds many values in order, under one name — perfect for an inventory, a party roster, or anything else that grows over the course of a game. Strings turn out to behave a lot like lists too, once you know what to look for.

## The Idea

A list is written with square brackets, values separated by commas:

```python
inventory = ["sword", "shield", "potion"]
```

Access one item by its **index** — counting starts at `0`:

```python
inventory[0]   # "sword"
```

Loop through every item with `for`:

```python
for item in inventory:
    print(item)
```

Add to a list with `.append()`, and check how many items it holds with `len()`. Strings support indexing and slicing the same way lists do — `"dragon"[0]` is `"d"`, and `"dragon"[0:3]` is `"dra"` — but unlike lists, you can't change a string in place; you can only build a new one.

````python
# A worked example — Zara's starting inventory

inventory = ["sword", "shield", "potion"]

print("Starting inventory:")
for item in inventory:
    print("-", item)

inventory.append("torch")
print(f"\nAfter finding a torch, you're carrying {len(inventory)} items:")
for item in inventory:
    print("-", item)
````

*Watch out for:* indexing starts at `0`, not `1` — `inventory[3]` on a 3-item list is an `IndexError`, since the valid indexes are `0`, `1`, and `2`. And strings are **immutable**: `name[0] = "Z"` will error, even though the equivalent works fine on a list.
