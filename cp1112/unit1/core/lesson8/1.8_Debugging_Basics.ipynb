# 1.8 — Debugging Basics

Every programmer, at every level, spends serious time fixing code that doesn't work. That's not a sign something's gone wrong with *you* — it's just what programming actually is. Today's about doing it efficiently: reading error messages properly, and tracing through code by hand when the bug isn't obvious.

## The Idea

Python's error messages (**tracebacks**) tell you three things: what kind of error it is, what line it happened on, and often a description of exactly what went wrong. Read them **bottom to top** — the last line names the error type and gives the clearest description; the lines above show you how the program got there.

Common error types you've already met:
- **`NameError`** — you used a variable that doesn't exist (often a typo, or forgetting to define it first)
- **`TypeError`** — you tried to combine or use values in a way their types don't support (like `"5" + 5`)
- **`IndexError`** — you asked for a list or string index that doesn't exist

When the bug isn't obvious from the error alone, add temporary `print()` statements to show what a variable actually contains at each step — this is called **print-statement debugging**, and it's the single most useful debugging skill you can build right now.

```python
# A worked example — reading a traceback and fixing the bug

# Buggy version (this cell will error on purpose):
# print(character_nam)

# The traceback would end with:
# NameError: name 'character_nam' is not defined

# Fixed version:
character_name = "Zara"
print(character_name)
```
