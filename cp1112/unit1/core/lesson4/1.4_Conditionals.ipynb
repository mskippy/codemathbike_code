# 1.4 — Conditionals

Every program you've written so far runs the exact same way every time. Conditionals change that — your code can finally make decisions, which means Zara's story can actually branch.

## The Idea

`if` runs a block of code only when a condition is `True`. Add `elif` for more conditions to check, and `else` to catch everything else:

```python
if health <= 0:
    print("Game over.")
elif health < 20:
    print("Critically wounded!")
else:
    print("Still standing.")
```

Python checks conditions top to bottom and runs the **first** one that's `True` — the rest are skipped, even if they'd also be `True`. Combine conditions with `and` (both must be true), `or` (at least one must be true), and `not` (flips a `bool`).

*Watch out for:* `=` assigns a value; `==` compares two values. Mixing them up is one of the most common bugs in this lesson — and confusingly, `if health = 100:` isn't just wrong, it's a `SyntaxError`, so at least Python won't let it run silently.

````python
# A worked example — checking if Zara can still fight

health = 15

if health <= 0:
    print("Zara has fallen.")
elif health < 20:
    print("Zara is critically wounded, but still fighting!")
else:
    print("Zara is in good shape.")
````
