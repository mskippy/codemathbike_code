# 1.2 — Input/Output & f-strings

So far every value in your programs has been typed directly into the code. Today your programs start listening: `input()` lets a person type something back, and f-strings are the clean way to weave that (or any variable) into a sentence.

## The Idea

`input("prompt")` pauses your program, shows the prompt, and waits for the player to type something and hit Enter. Whatever they type comes back as a **string** — always, even if they typed a number.

```python
player_name = input("What's your name, adventurer? ")
```

An **f-string** lets you drop variables straight into a sentence — put an `f` right before the opening quote, and wrap each variable in `{ }`:

```python
print(f"Welcome, {player_name}!")
```

If you need a number instead of text, wrap `input()` in `int()` or `float()` to convert it:

```python
starting_gold = int(input("How much gold do you start with? "))
```

````python
# A worked example — greeting Zara at the start of the adventure

player_name = input("What's your name, adventurer? ")
starting_gold = int(input("How much gold do you start with? "))

print(f"Welcome, {player_name}!")
print(f"You set out with {starting_gold} gold in your pocket.")
````

*Watch out for:* `input()` **always** hands back a string, even if the player types `100`. Try to add two un-converted inputs together and you'll get string concatenation (`"10" + "5"` gives `"105"`, not `15`) instead of addition — wrap numeric input in `int()` or `float()` the moment you read it.
