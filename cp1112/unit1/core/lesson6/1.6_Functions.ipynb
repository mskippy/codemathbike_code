# 1.6 — Functions

You've been writing the same *kind* of code over and over — reading input, calculating damage, printing a message. Functions let you package that up once, give it a name, and reuse it — the first real step toward organizing a program instead of just writing it top to bottom.

## The Idea

`def` creates a function. Anything in parentheses is a **parameter** — a value the function needs from whoever calls it. `return` sends a value back out; a function with no `return` gives back `None`.

```python
def calculate_damage(attack, defense):
    return attack - defense

result = calculate_damage(18, 5)   # result is now 13
```

A variable created *inside* a function only exists inside that function — that's called **scope**. If you need a value to exist outside the function, `return` it and assign it to a variable in the outer code, like `result` above.

````python
# A worked example — a reusable damage calculator

def calculate_damage(attack, defense):
    return attack - defense

zara_damage = calculate_damage(18, 5)
finn_damage = calculate_damage(12, 3)

print(f"Zara deals {zara_damage} damage.")
print(f"Old Finn deals {finn_damage} damage.")
````

*Watch out for:* `print()`-ing inside a function shows something on screen, but doesn't give the calling code anything to work with afterward. If you need to use the result later (store it, compare it, pass it to something else), you need `return`, not just `print()`.
