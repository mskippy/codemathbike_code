## Tier 1 — Personalize it

A **variable** is just a labeled box that holds a value. `print()` is how you get Python to actually show you something on screen. Put a string (text in quotes) inside a variable, then build a sentence around it with `+`.

```python
favorite_snack = "popcorn"
print("My favorite snack is " + favorite_snack + ".")
```

*Side note: you'll also see people write this as an f-string — `print(f"My favorite snack is {favorite_snack}.")` — same result, different syntax. Either is fine.*

## Tier 2 — Add more

Nothing new syntax-wise here — just more of the same. Each `print()` is its own line of output, and you can have as many variables as you want before you print them.

```python
favorite_snack = "popcorn"
hobby = "rock climbing"

print("My favorite snack is " + favorite_snack + ".")
print("In my spare time I like " + hobby + ".")
```

## Tier 3 — Make it interactive

`input()` pauses your program and waits for whoever's running it to type something and hit Enter. Whatever they type comes back as a string, which you can store in a variable just like before.

```python
snack = input("What's your favorite snack? ")
print("Nice, " + snack + " is a great choice.")
```

## Tier 4 — Add a decision

`if` / `else` lets your program choose between two (or more) things to do, based on a condition — something that's either `True` or `False`. Here, the condition is whether the typed answer matches what we're checking for.

```python
answer = input("Do you like pineapple on pizza? (yes/no) ")

if answer == "yes":
    print("Bold choice.")
else:
    print("Fair enough.")
```

## Tier 5 — Add a loop

A `for` loop repeats a block of code once for each item in a list — no need to write a separate `print()` for every single one.

```python
facts = ["I have a dog", "I like hiking", "I've never left North America"]

for fact in facts:
    print(fact)
```

## Tier 6 — Go further

This tier is intentionally open — it's less about one specific new bit of syntax, more about what you do with what you already know (or what you're curious enough to look up). A few directions people sometimes take it, if you want *ideas* and not answers:

- Combine a loop **and** a decision — print something different depending on what's in the list
- Turn a repeated chunk of code into your own function with `def`
- Ask more than one question with `input()` and use all the answers together
- Try something with lists, dictionaries, or formatting that hasn't come up here

Whatever you pick, the goal's the same as every tier: something that actually runs.
