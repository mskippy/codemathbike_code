```python
import json
import os

def save_data(data, filename):
    """Write the whole data set to a JSON file, replacing what was there."""
    with open(filename, "w") as f:
        json.dump(data, f, indent=2)

def load_data(filename):
    """Read a JSON file and return the data it holds."""
    with open(filename) as f:
        return json.load(f)

def load_or_start(filename):
    """Load the file if it exists. On a first run, return an empty dictionary."""
    if os.path.exists(filename):
        return load_data(filename)
    return {}

import io, contextlib

def check_case(label, got, expected):
    """Print PASS or FAIL for one test case, without stopping the program."""
    if got == expected:
        print("PASS: " + label)
    else:
        print("FAIL: " + label + ". Expected " + repr(expected) + ", got " + repr(got))

def printed_pass(test_function, *args):
    """Run a test and return True if it printed PASS."""
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        test_function(*args)
    return "PASS" in buf.getvalue()
```

## Part 1 — The round-trip test

The most important test for a saved app is: **save it, load it, compare.** `assert` stops the program with an error if the condition is False. The message after the comma tells you what failed.

Run this example. It should print PASS.

```python
def test_round_trip():
    original = {"apple": {"qty": 4, "price": 0.50}}
    save_data(original, "test_round_trip.json")
    loaded = load_data("test_round_trip.json")
    assert loaded == original, "Data changed after save and load"
    print("PASS: round trip")

test_round_trip()
```

Note: this test uses its own file name. Your tests should never touch your real save file.
