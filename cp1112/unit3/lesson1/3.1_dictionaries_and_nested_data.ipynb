## Part 1 — Why not just a list?

Here's how you'd track two things about a shopping list with two lists:

```python
names = ["Apple", "Bread", "Milk"]
qtys  = [4, 1, 2]

# How many breads? You have to remember that Bread is at position 1.
print(qtys[names.index("Bread")])
```

That works, but the two lists have to stay in step. If you add a name and forget the quantity, everything after that point is wrong, and Python won't warn you.

A **dictionary** keeps each name and its values **together**. Keep going to Part 2.

## Part 2 — One record is a dictionary

A dictionary stores **key: value** pairs inside `{ }`.

| What you want | Code | What it does |
|---|---|---|
| Read a value | `apple["qty"]` | Gives `4`. Error if the key isn't there. |
| Change a value | `apple["qty"] = 3` | Replaces the value for that key |
| Add a new key | `apple["colour"] = "red"` | Adds a new key and value |
| Check a key exists | `"price" in apple` | `True` or `False` |
| Read safely | `apple.get("colour", "not set")` | Gives the default if the key is missing |

```python
apple = {"name": "Apple", "qty": 4, "price": 0.50}

print(apple["qty"])                    # 4
apple["qty"] = 3                       # change a value
print(apple["qty"])                    # 3
print("price" in apple)                # True
print(apple.get("colour", "not set"))  # not set
```

**Rule of thumb:** use `[ ]` when you know the key is there. Use `.get()` when it might not be.

## Part 4 — A dictionary of records (look up by key)

If you want to find a record by its name, make the name the key. Compare these two:

- A **list** of records: you loop through to find `"Bread"`.
- A **dictionary** of records: you ask for `inventory["bread"]` directly.

```python
inventory = {
    "apple": {"qty": 4, "price": 0.50},
    "bread": {"qty": 1, "price": 2.25},
}

print(inventory["bread"]["price"])     # 2.25

# .items() gives you each key and its value together
for name, details in inventory.items():
    print(name, details["qty"])
```

## Part 5 — Nested data (a dictionary inside a dictionary, with lists inside)

Real data often has layers. A gradebook might look like this: each student has a dictionary of subjects, and each subject has a **list** of marks.

```python
gradebook = {
    "Priya": {"math": [88, 92, 79], "science": [75, 81]},
    "Sam":   {"math": [65, 70, 72], "science": [90, 85]},
}

print(gradebook["Priya"])              # the whole record for Priya
print(gradebook["Priya"]["math"])      # [88, 92, 79]
print(gradebook["Priya"]["math"][1])   # 92: the second math mark
```

**How to read it:** go left to right, one layer at a time.

`gradebook["Priya"]["math"][1]`
1. the student (`"Priya"`), then
2. the subject (`"math"`), then
3. the position in that list (`1`, which is the **second** mark, since counting starts at 0).

## Part 6 — Safe looping

Two things trip people up when they loop over a dictionary:

1. **Changing a dictionary while you loop over it** can break the loop. If you need to add or remove entries, first make a copy of the keys, then loop over the copy.
2. **Two records with the same key** overwrite each other with no error. Decide what your key is before you start. For example, if two students can have the same name, use a student ID as the key.

```python
# Safe: loop over a copy of the keys
for name in list(gradebook.keys()):
    if name == "Sam":
        gradebook[name]["math"].append(100)   # changing values is fine

print(gradebook["Sam"]["math"])
```
