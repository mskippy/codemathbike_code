```python
import io, contextlib

def check_case(label, got, expected):
    """Print PASS or FAIL for one test case, without stopping the program."""
    if got == expected:
        print("PASS: " + label)
    else:
        print("FAIL: " + label + ". Expected " + repr(expected) + ", got " + repr(got))

def printed_pass(test_function, *args):
    """Run a test and return True if it printed PASS."""
    buf = io.StringIO()
    with contextlib.redirect_stdout(buf):
        test_function(*args)
    return "PASS" in buf.getvalue()
```

## Part 1 — Data in, data out

These two functions don't call `input()` or `print()`. They take data in and return data out. The menu does the talking. That's what makes them easy to test and easy to reuse.

```python
def add_item(inventory, name, qty, price):
    """Return the inventory with this item added, or updated if it's already there."""
    inventory[name] = {"qty": qty, "price": price}
    return inventory

def total_value(inventory):
    """Return the total value of everything in stock."""
    total = 0
    for details in inventory.values():
        total += details["qty"] * details["price"]
    return total

stock = {}
stock = add_item(stock, "apple", 4, 0.50)
stock = add_item(stock, "bread", 1, 2.25)
print(total_value(stock))   # 4 * 0.50 + 1 * 2.25 = 4.25
```
