## Part 1 — Save, then load

When your program closes, its variables disappear. A file keeps the data between runs. JSON is a text format that stores dictionaries and lists, and Python reads it back into the same shapes.

These three functions do the work. Run the cell, then read the example.

```python
import json
import os

def save_data(data, filename):
    """Write the whole data set to a JSON file, replacing what was there."""
    with open(filename, "w") as f:
        json.dump(data, f, indent=2)

def load_data(filename):
    """Read a JSON file and return the data it holds."""
    with open(filename) as f:
        return json.load(f)

def load_or_start(filename):
    """Load the file if it exists. On a first run, return an empty dictionary."""
    if os.path.exists(filename):
        return load_data(filename)
    return {}
```

**Worked example.** Save a gradebook, look at the file that was written, then load it back.

```python
gradebook = {"Priya": {"math": [88, 92]}, "Sam": {"math": [65, 70]}}
save_data(gradebook, "demo_gradebook.json")

print(open("demo_gradebook.json").read())   # this is the file, as text
loaded = load_data("demo_gradebook.json")
print(loaded == gradebook)                  # True
```

Look at the printed text. It's your data, written with double quotes. That's what JSON looks like.

## Part 2 — Two rules that catch people out

**Rule 1: keys always come back as text.** **Rule 2: only some types can be saved.** Dictionaries, lists, text, numbers, `true`/`false` and `None` are fine. Sets and custom objects are not.

Run this and read the output carefully.

```python
save_data({1: "Priya", 2: "Sam"}, "demo_ids.json")
print(load_data("demo_ids.json"))    # look at the keys: are they numbers, or text?

try:
    save_data({"tags": {"red", "blue"}}, "demo_bad.json")   # a set can't be saved
except TypeError as error:
    print("Couldn't save:", error)
```
